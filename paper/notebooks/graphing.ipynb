edit and run. saved figures land in ../paper/figures/ and flow into the paper.

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
plt.style.use('loom.mplstyle')

# the style turns text.usetex on. test-render once and fall back if latex or dvipng is missing.
try:
    _probe = plt.figure()
    _probe.text(0.5, 0.5, r'$\sin x$')
    _probe.canvas.draw()
except Exception as exc:
    plt.rcParams['text.usetex'] = False
    print(f'latex unavailable, falling back to usetex=False ({type(exc).__name__})')
finally:
    plt.close('all')
```

```python
# example. rename and save into the paper figures dir.
fig, ax = plt.subplots()
x = np.linspace(0, 2 * np.pi, 200)
ax.plot(x, np.sin(x), label=r'$\sin x$')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.legend()
fig.savefig('../paper/figures/example.png')
```
